# Parte A — Questões de múltipla escolha 

Exercício 1 — Objetos, encapsulamento e interface  [Médio–difícil] 
Considere uma classe ContaBancaria em que o atributo __saldo é privado e as operações depositar(), sacar() 
e consultar_saldo() são públicas. Qual alternativa interpreta melhor essa modelagem segundo os princípios 
estudados? 
 
 B) O encapsulamento permite ocultar a representação interna do saldo e disponibilizar uma interface de 
alto nível para manipulação do objeto.

Exercício 2 — Herança e polimorfismo  [Médio–difícil] 
Uma aplicação possui a classe Funcionario, com o método calcular_bonus(). As classes Vendedor e Gerente 
herdam de Funcionario e redefinem calcular_bonus() de formas diferentes. Ao percorrer uma lista de objetos 
e executar funcionario.calcular_bonus(), cada objeto produz o resultado adequado ao seu tipo. Quais 
conceitos aparecem de forma mais direta? 

C) Herança e polimorfismo.

Exercício 3 — Atributo de instância x atributo de classe  [Médio–difícil] 
Observe o código: class Produto: contador = 0; def __init__(self, nome): Produto.contador += 1; self.nome = 
nome. Depois são criados três objetos Produto. Qual afirmação é correta? 

B) Produto.contador será compartilhado e terá valor 3 após as três instanciações.

# Exercício 5 — Conta simples com encapsulamento  [Fácil] 

Crie uma classe Conta que mantenha o saldo protegido contra alteração direta. 
Requisitos: 
• O construtor recebe o nome do titular e inicia o saldo em 0. 

• Use __saldo como atributo privado. 

• Implemente depositar(valor), permitindo apenas valores positivos. 

• Implemente consultar_saldo(), retornando o saldo atual.
 
• Crie uma conta, faça dois depósitos e mostre o saldo.

# Exercício 7 — Veículo com velocidade controlada  [Média] 
Modele um veículo inspirado no exemplo Automovel da aula, mas com regras completas de aceleração e 
frenagem. 

Requisitos: 

• A classe Veiculo deve ter placa, velocidade_max e velocidade_atual; torne os três atributos privados. 

• velocidade_atual começa em 0. 

• Crie getters para placa e velocidade_max. 

• Crie um setter para velocidade_max que aceite apenas valores positivos e nunca deixe a nova máxima 
menor que a velocidade atual. 

• acelerar(incremento=10) aumenta a velocidade sem ultrapassar a máxima. 

• frear(decremento=10) reduz a velocidade sem ficar negativa. 

• Implemente __str__ para mostrar placa, velocidade atual e máxima. 

• Demonstre o funcionamento com várias acelerações e frenagens.

# Exercício 8 — Funcionários e cálculo de bônus  [Média] 

Crie uma hierarquia para calcular bônus de funcionários de uma empresa. 

Requisitos: 

• Funcionario possui nome e salario; implemente __str__. 

• Funcionario.calcular_bonus() retorna 5% do salário. 

• Vendedor herda de Funcionario, acrescenta vendas e redefine calcular_bonus() para 5% do salário + 2% 
das vendas. 

• Gerente herda de Funcionario, acrescenta equipe e redefine calcular_bonus() para 10% do salário + R$ 
100 por integrante da equipe. 

• Crie pelo menos um objeto de cada tipo, coloque-os em uma única lista e exiba nome, tipo e bônus 
usando o mesmo método calcular_bonus(). 
 

# Exercício 9 — Produtos e contador de objetos  [Média] 

Crie uma classe Produto para um catálogo de e-commerce, explorando atributo de classe e encapsulamento. 

Requisitos: 

• Crie o atributo de classe contador iniciado em 0. 

• Cada produto deve receber automaticamente um id sequencial a partir do contador. 

• Use atributos privados para id, nome e preco. 

• Crie getters para id, nome e preco. 

• Crie set_preco(novo_preco), aceitando apenas valores maiores que zero. 

• Implemente aplicar_desconto(percentual), validando percentual entre 0 e 100. 

• Implemente __str__. 

• Crie três produtos e, ao final, mostre quantos objetos foram criados usando Produto.contador.

# Exercício 10 — Mini sistema de pedidos de e-commerce  [Difícil] 

Desenvolva um pequeno sistema orientado a objetos para pedidos de um e-commerce. O objetivo é integrar a 
maior parte dos conceitos trabalhados na aula em um único problema. 

Requisitos: 

• Crie uma classe Produto com id automático (atributo de classe), nome e preço privados; inclua getters, 
setter validado para preço e __str__. 

• Crie ProdutoFisico herdando de Produto, com peso, e ProdutoDigital herdando de Produto, com 
tamanho_mb. 

• Em Produto, crie calcular_frete() retornando 0. Redefina-o em ProdutoFisico para retornar R$ 5,00 + R$ 
2,00 por kg. ProdutoDigital deve manter frete igual a 0. Essa chamada comum deve demonstrar 
polimorfismo. 

• Crie uma classe Cliente com nome e email privados, getters e __str__. 

• Crie uma classe Pedido com número automático (atributo de classe), cliente, lista de produtos e status 
privado. 

• Pedido deve oferecer adicionar_produto(produto), calcular_total(), calcular_frete_total(), fechar_pedido() 
e __str__. 

• Depois de fechado, o pedido não deve aceitar novos produtos. Essa regra deve ser controlada 
internamente pela classe. 

• calcular_total() deve somar os preços dos produtos. calcular_frete_total() deve chamar calcular_frete() de 
cada item sem testar se ele é físico ou digital. 

• No programa principal, crie um cliente, ao menos três produtos (misturando físicos e digitais), monte um 
pedido, exiba o resumo, feche-o e tente adicionar mais um produto para comprovar a regra
