# Exercício 7 — Veículo com velocidade controlada  [Média] 
Modele um veículo inspirado no exemplo Automovel da aula, mas com regras completas de aceleração e 
frenagem. 

Requisitos: 

• A classe Veiculo deve ter placa, velocidade_max e velocidade_atual; torne os três atributos privados. 

• velocidade_atual começa em 0. 

• Crie getters para placa e velocidade_max. 

• Crie um setter para velocidade_max que aceite apenas valores positivos e nunca deixe a nova máxima 
menor que a velocidade atual. 

• acelerar(incremento=10) aumenta a velocidade sem ultrapassar a máxima. 

• frear(decremento=10) reduz a velocidade sem ficar negativa. 

• Implemente __str__ para mostrar placa, velocidade atual e máxima. 

• Demonstre o funcionamento com várias acelerações e frenagens.

```python


class Veiculo:
    def __init__ (self, placa, velocidade_max, velocidade_atual=0):
        self.__placa = placa
        self.__velocidade_max = velocidade_max
        self.__velocidade_atual = velocidade_atual

    def mostrar_placa(self):
        return self.__placa

    def velocidade_max(self):
        return self.__velocidade_max

    def set_velocidade_max(self, nova_max):
        if nova_max > 0 and nova_max >= self.__velocidade_atual:
            self.__velocidade_max = nova_max
        else:
            print("Velocidade máxima inválida!")

    def acelerar(self, incremento=10):
        self.__velocidade_atual = self.__velocidade_atual + incremento
        if self.__velocidade_atual > self.__velocidade_max:
            self.__velocidade_atual = self.__velocidade_max

    def frear(self, decremento=10):
        self.__velocidade_atual = self.__velocidade_atual - decremento
        if self.__velocidade_atual < 0:
            self.__velocidade_atual = 0

    def __str__(self):
        return (f"Placa: {self.__placa}, Velocidade Atual: {self.__velocidade_atual} km/h, Máxima: {self.__velocidade_max} km/h")



carro = Veiculo("ABC-1234", 100)
print(carro)

carro.acelerar(50)
print(carro)

carro.acelerar(70) 
print(carro)

carro.frear(30)
print(carro)

carro.frear(100) 
print(carro)
        
        
```

# Exercício 8 — Funcionários e cálculo de bônus  [Média] 

Crie uma hierarquia para calcular bônus de funcionários de uma empresa. 

Requisitos: 

• Funcionario possui nome e salario; implemente __str__. 

• Funcionario.calcular_bonus() retorna 5% do salário. 

• Vendedor herda de Funcionario, acrescenta vendas e redefine calcular_bonus() para 5% do salário + 2% 
das vendas. 

• Gerente herda de Funcionario, acrescenta equipe e redefine calcular_bonus() para 10% do salário + R$ 
100 por integrante da equipe. 

• Crie pelo menos um objeto de cada tipo, coloque-os em uma única lista e exiba nome, tipo e bônus 
usando o mesmo método calcular_bonus(). 
 

```python
class Funcionario:
    def __init__(self, nome, salario):
        self.nome = nome
        self.salario = salario

    def calcular_bonus(self):
        return self.salario * 0.05

    def __str__(self):
        return f"Nome: {self.nome} | Salário: R$ {self.salario:.2f}"


class Vendedor(Funcionario):
    def __init__(self, nome, salario, vendas):
        super().__init__(nome, salario)
        self.vendas = vendas

    def calcular_bonus(self):
        return (self.salario * 0.05) + (self.vendas * 0.02)


class Gerente(Funcionario):
    def __init__(self, nome, salario, equipe):
        super().__init__(nome, salario)
        self.equipe = equipe

    def calcular_bonus(self):
        return (self.salario * 0.10) + (len(self.equipe) * 100)


f1 = Funcionario("João", 2000.0)
v1 = Vendedor("Maria", 2500.0, 10000.0)
g1 = Gerente("Carlos", 5000.0, ["Ana", "Pedro", "Lucas"])

lista_funcionarios = [f1, v1, g1]

for f in lista_funcionarios:
    tipo = type(f).__name__
    bonus = f.calcular_bonus()
    print(f"Nome: {f.nome}, Tipo: {tipo}, Bônus: R$ {bonus:.2f}")
```

• Cada produto deve receber automaticamente um id sequencial a partir do contador. 

• Use atributos privados para id, nome e preco. 

• Crie getters para id, nome e preco. 

• Crie set_preco(novo_preco), aceitando apenas valores maiores que zero.

• Implemente __str__. 

• Crie três produtos e, ao final, mostre quantos objetos foram criados usando Produto.contador. 

```python
class Produto:
    contador = 0

    def __init__(self, nome, preco):
        Produto.contador = Produto.contador + 1
        self.__id = Produto.contador
        self.__nome = nome
        self.__preco = preco

    def id(self):
        return self.__id

    def nome(self):
        return self.__nome

    def preco(self):
        return self.__preco

    def preco(self, novo_preco):
        if novo_preco > 0:
            self.__preco = novo_preco
        else:
            print("Preço deve ser maior que zero!")

    def aplicar_desconto(self, percentual):
        if percentual >= 0 and percentual <= 100:
            desconto = self.__preco * (percentual / 100)
            self.__preco = self.__preco - desconto
        else:
            print("Percentual de desconto inválido!")

    def __str__(self):
        return f"ID: {self.__id}, Nome: {self.__nome} , Preço: R$ {self.__preco:.2f}"



p1 = Produto("Camiseta", 50.0)
p2 = Produto("Calça", 120.0)
p3 = Produto("Tênis", 200.0)

print(p1)
print(p2)
print(p3)

p1.aplicar_desconto(10) 
print("Após desconto na camiseta:", p1)

print("Total de produtos criados:", Produto.contador)
```

maior parte dos conceitos trabalhados na aula em um único problema.

• Crie uma classe Pedido com número automático (atributo de classe), cliente, lista de produtos e status 
privado. 

• Pedido deve oferecer adicionar_produto(produto), calcular_total(), calcular_frete_total(), fechar_pedido() 
e __str__. 

• Depois de fechado, o pedido não deve aceitar novos produtos. Essa regra deve ser controlada 
internamente pela classe. 

• calcular_total() deve somar os preços dos produtos. calcular_frete_total() deve chamar calcular_frete() de 
cada item sem testar se ele é físico ou digital. 

• No programa principal, crie um cliente, ao menos três produtos (misturando físicos e digitais), monte um 
pedido, exiba o resumo, feche-o e tente adicionar mais um produto para comprovar a regra

A entrega deve ser feita diretamente no repositório GitHub.

```python
class Produto:
    contador = 0

    def __init__(self, nome, preco):
        Produto.contador = Produto.contador + 1
        self.__id = Produto.contador
        self.__nome = nome
        self.__preco = preco
        
        
    def id(self):
        return (self.__id)

    def get_nome(self):
        return (self.__nome)

    def get_preco(self):  
        return (self.__preco)

    def set_preco(self, novo_preco): 
        if novo_preco > 0:
            self.__preco = novo_preco
        else:
            print("Preço inválido!")

    def calcular_frete(self):
        return (0.0)

    def __str__(self):
        return (f"[{self.__id}] {self.__nome} R$ {self.__preco}")


class ProdutoFisico(Produto):
    def __init__(self, nome, preco, peso):
        super().__init__(nome, preco)
        self.peso = peso

    def calcular_frete(self):
        return (5.0 + (self.peso * 2.0))


class ProdutoDigital(Produto):
    def __init__(self, nome, preco, tamanho_mb):
        super().__init__(nome, preco)
        self.tamanho_mb = tamanho_mb

    def calcular_frete(self):
        return (0.0)


class Cliente:
    def __init__(self, nome, email):
        self.__nome = nome
        self.__email = email

    def nome(self):
        return (self.__nome)

    def email(self):
        return (self.__email)

    def __str__(self):
        return (f"Cliente: {self.__nome} ({self.__email})")


class Pedido:
    contador = 0

    def __init__(self, cliente):
        Pedido.contador = Pedido.contador + 1
        self.numero = Pedido.contador
        self.cliente = cliente
        self.produtos = []
        self.__status = "Aberto"

    def adicionar_produto(self, produto):
        if self.__status == "Aberto":
            self.produtos.append(produto)
            print(f"Produto '{produto.get_nome()}' adicionado ao pedido.")
        else:
            print("Não é possível adicionar produtos: o pedido já está fechado!")

    def calcular_total(self):
        total = 0.0
        for prod in self.produtos:
            total = total + prod.get_preco()  
        return (total)

    def calcular_frete_total(self):
        frete_total = 0.0
        for prod in self.produtos:
            frete_total = frete_total + prod.calcular_frete()
        return (frete_total)

    def fechar_pedido(self):
        self.__status = "Fechado"
        print("Pedido fechado com sucesso!")

    def __str__(self):
        resumo = (f"--- PEDIDO Nº {self.numero} ---")
        resumo += (f"{self.cliente}")
        resumo += (f"Status: {self.__status}")

        resumo += ("Itens:")
        for p in self.produtos:
            resumo += (f"  - {p}")
        resumo += (f"Total Produtos: R$ {self.calcular_total()}")
        resumo += (f"Total Frete: R$ {self.calcular_frete_total()}")
        resumo += (f"Valor Final: R$ {self.calcular_total() + self.calcular_frete_total()}")
        return resumo



c1 = Cliente("Ana Silva", "ana@email.com")

prod1 = ProdutoFisico("Livro impresso", 40.0, 1.5)    
prod2 = ProdutoDigital("E-book Python", 25.0, 15)    
prod3 = ProdutoFisico("Cadeira Gamer", 800.0, 12.0)   

pedido1 = Pedido(c1)

print("--- Adicionando produtos ---")
pedido1.adicionar_produto(prod1)
pedido1.adicionar_produto(prod2)
pedido1.adicionar_produto(prod3)

print("--- Resumo do Pedido ---")
print(pedido1)

print("--- Fechando o pedido ---")
pedido1.fechar_pedido()

print("--- Tentando adicionar outro produto após fechar ---")
prod4 = ProdutoDigital("Curso em Vídeo", 100.0, 500)
pedido1.adicionar_produto(prod4)
```
