```python
def busca_seq (l,n):
     for i in range (len(l)):
         if l [i] == n:
              return i
         return -1

def busca_seq2 (l,n):
    i = 0
    while i <= len(l):
        if  l[i] == n:
            return i
        i = i + 1
        return -1
```

```python
l = [5, 10, 20, 21, 3, 15, 1, 2, 18, 26]
numero = int (input("Digite o valor a ser pesquisado"))
indice = busca_seq (l, numero)
if indice == -1:
    print ("O número n foi encontrado!")
else:
    print (f"O {numero} foi encontrado na posição {indice}")
```
