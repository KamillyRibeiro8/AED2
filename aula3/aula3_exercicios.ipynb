4. (PRÁTICA): Implemente uma função em Python chamada buscar_cliente(nome, lista_clientes) que
realiza busca sequencial em uma lista de nomes de clientes. Essa função deve retornar o índice onde o nome foi
encontrado ou -1 caso o nome não esteja na lista_clientes. Em seguida, escreva um pequeno código de
exemplo que utilize essa função para buscar por pelo menos um nome que exista e um que não exista na lista,
exibindo mensagens adequadas.

```python
def buscar_cliente(nome, lista_clientes):
    for i in range(len(lista_clientes)):
        if lista_clientes[i] == nome:
            return i
    return -1

clientes = ["Ana", "Carlos", "Mariana", "João"]

nome_existente = "Mariana"
indice = buscar_cliente(nome_existente, clientes)
if indice != -1:
    print(f"O cliente '{nome_existente}' foi encontrado no índice {indice}.")
else:
    print(f"O cliente '{nome_existente}' não foi encontrado.")

nome_inexistente = "Pedro"
indice = buscar_cliente(nome_inexistente, clientes)
if indice != -1:
    print(f"O cliente '{nome_inexistente}' foi encontrado no índice {indice}.")
else:
    print(f"O cliente '{nome_inexistente}' não foi encontrado.")
```

5. (PRÁTICA): Crie uma função conta_ocorrencias(valor, lista) em Python que percorre uma lista e conta
quantas vezes um determinado valor aparece. Utilize busca sequencial (varrendo a lista) para realizar a
contagem. Por exemplo, conta_ocorrencias(3, [1,3,5,3,7]) deve retornar 2, pois o valor 3 aparece duas
vezes na lista. Inclua casos de teste chamando a função com diferentes valores e listas.

```python
def conta_ocorrencias(valor, lista):
    contador = 0
    for elemento in lista:
        if elemento == valor:
            contador += 1
    return contador

lista1 = [1, 3, 5, 3, 7]
resultado1 = conta_ocorrencias(3, lista1)
print(f"O valor 3 aparece {resultado1} vezes na lista {lista1}.")


lista2 = [10, 20, 30, 40]
resultado2 = conta_ocorrencias(50, lista2)
print(f"O valor 50 aparece {resultado2} vezes na lista {lista2}.")


lista3 = ["massa", "queijo", "massa", "molho", "massa"]
resultado3 = conta_ocorrencias("massa", lista3)
print(f"O valor 'massa' aparece {resultado3} vezes na lista {lista3}.")
```

6. (PRÁTICA): Suponha que precisamos remover um elemento específico de uma lista não ordenada. Escreva
uma função Python remove_elemento(lista, valor) que remove a primeira ocorrência de valor na lista,
mantendo os demais elementos e retornando o valor removido. A função deve retornar None caso o valor não
seja encontrado ou a lista esteja vazia. Dica: Utilize a busca sequencial para encontrar a posição do elemento
e, se encontrada, deslize os elementos posteriores para “preencher” o espaço removido (ou use métodos
próprios da lista Python). Demonstre o funcionamento da função com um exemplo.

```python
def remove_elemento(lista, valor):
   
    if not lista:
        return None
    
    indice_encontrado = -1
    for i in range(len(lista)):
        if lista[i] == valor:
            indice_encontrado = i
            break 
            
    if indice_encontrado == -1:
        return None
    
    return lista.pop(indice_encontrado)

minha_lista = [10, 25, 40, 25, 50]
print("Lista original:", minha_lista)

removido = remove_elemento(minha_lista, 25)
print(f"Valor removido: {removido}")
print("Lista após remoção:", minha_lista)

removido_inexistente = remove_elemento(minha_lista, 99)
print(f"Valor removido: {removido_inexistente}")

removido_vazia = remove_elemento([], 10)
print(f"Resultado com lista vazia: {removido_vazia}")
```

# Projeto 1: Encontrando seu Filme Favorito
Contexto: Imagine um sistema de recomendação de filmes simples onde os filmes são armazenados em uma
lista não-ordenada.

Problema: Implementar uma função que permita ao usuário buscar um filme pelo título nessa lista.

Indicação de Solução: Crie uma lista de títulos de filmes e uma função que use a busca sequencial para
encontrar um título inserido pelo usuário. Se o filme for encontrado, exiba uma mensagem informando o título e
sua posição (índice) na lista. Se não for encontrado, exiba uma mensagem informando que o filme não está no
catálogo.


```python
def buscar_filme(titulo_procurado, catálogo):
   
    for i in range(len(catálogo)):
        if catálogo[i].lower() == titulo_procurado.lower():
            return i
    return -1

catalogo_filmes = [
    "Matrix",
    "O Senhor dos Anéis",
    "Inception",
    "Interstellar",
    "De Volta para o Futuro"
]

filme_usuario = input("Digite o nome do filme que deseja buscar: ")

posicao = buscar_filme(filme_usuario, catalogo_filmes)

if posicao != -1:
    print(f"O filme '{catalogo_filmes[posicao]}' foi encontrado no catálogo no índice {posicao}.")
else:
    print(f"O filme '{filme_usuario}' não está disponível no catálogo.")
```

# Projeto 2: Verificando a Lista de Convidados
Contexto: Desenvolver um programa para gerenciar a lista de convidados de um evento.

Problema: Implementar uma funcionalidade que permita verificar se um nome específico está na lista de
convidados.

Indicação de Solução: Crie uma lista de nomes de convidados. Implemente uma função que receba um nome
como entrada e use a busca sequencial para verificar se esse nome existe na lista. Exiba uma mensagem
indicando se o convidado está na lista ou não.

```python
def verificar_convidado(nome_procurado, lista_convidados):
   
    for convidado in lista_convidados:
        if convidado.lower() == nome_procurado.lower():
            return True
    return False

convidados = ["Lucas", "Beatriz", "Gabriel", "Fernanda", "Thiago"]

nome_busca = input("Digite o nome do convidado para verificar: ")

if verificar_convidado(nome_busca, convidados):
    print(f" Confirmado! {nome_busca.title()} está na lista de convidados.")
else:
    print(f" Acesso negado. {nome_busca.title()} NÃO está na lista de convidados.")
```
