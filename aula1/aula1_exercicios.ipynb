Exercício 1: Crie uma lista em Python chamada frutas com os seguintes elementos: "maçã", "banana",
"laranja". Em seguida, adicione a fruta "morango" ao final da lista. Por fim, imprima a lista completa.

```python
frutas = ["maçã", "banana","laranja"]
frutas.append("morango")
print(frutas)
```

Exercício 2: Crie uma pilha (usando uma lista Python) chamada livros. Adicione os livros "O Pequeno
Príncipe", "Dom Quixote" e "1984" nesta ordem. Em seguida, simule a remoção do livro do topo da pilha
e imprima o livro removido e a pilha restante.


```python
livros = ["O Pequeno Príncipe", "Dom Quixote", "1984"]
livro_removido = livros.pop(0)
print("Livro removido: ", livro_removido)
print("Pilha restante: ", livros)
```

Exercício 3: Crie uma fila (usando uma lista Python) chamada clientes. Adicione os clientes "Ana",
"Bruno" e "Carla" nesta ordem. Em seguida, simule a chegada de um novo cliente "Daniel" e a saída do
primeiro cliente da fila. Imprima o cliente que saiu e a fila restante.


```python
clientes = ["Ana", "Bruno", "Carla"]
clientes.append("Daniel")
cliente_removido = clientes.pop(0)
print("Desistente: ", cliente_removido)
print("Fila: ", clientes)
```

Exercício 4: Crie um dicionário em Python chamado contatos para armazenar o número de telefone
de duas pessoas: "Ana" com o telefone "1234-5678" e "Bruno" com o telefone "9876-5432". Em seguida,
adicione o contato de "Carlos" com o telefone "1122-3344". Por fim, imprima o número de telefone da
Ana.

```python
contatos = {"Ana" : "1234-5678", "Bruno" : "9876-5432"}
contatos["Carlos"] = "1122-3344"
print("Contatos: ", contatos)
print(contatos["Ana"])
```

Projeto 1: A Agenda de Contatos:
• Contexto: Você precisa criar um sistema simples para armazenar e gerenciar contatos (nome,
telefone e e-mail).
• Problema: Como você pode adicionar novos contatos, buscar um contato pelo nome e listar
todos os contatos?
• Indicação de como resolver: Pensar em uma estrutura que permita associar informações
(telefone e e-mail) a um nome de forma eficiente.
• Resolução do Projeto 1: Um dicionário em Python é uma excelente escolha para este projeto.
A chave pode ser o nome do contato, e o valor pode ser outro dicionário contendo o telefone e
o e-mail.


```python
agenda = {
    "Ana": {"telefone": "1234-5678", "email": "ana@email.com"},
    "Bruno": {"telefone": "9876-5432", "email": "bruno@email.com"}
}

agenda["Carlos"] = {"telefone": "1122-3344", "email": "carlos@email.com"}

print("Dados da Ana:", agenda["Ana"])

print("Agenda completa:", agenda)
```

Projeto 2: A Lista de Tarefas:
• Contexto: Você precisa criar uma lista de tarefas para a semana.
• Problema: Como você pode adicionar novas tarefas, marcar tarefas como concluídas e
visualizar as tarefas pendentes?
• Indicação de como resolver: Pensar em uma forma de manter uma sequência de itens que
podem ser adicionados e removidos.
• Resolução do Projeto 2: Uma lista em Python é ideal para essa situação. Podemos adicionar
novas tarefas ao final da lista, remover tarefas quando concluídas e simplesmente imprimir a
lista para visualizar as tarefas pendentes.

```python
tarefas = ["Estudar Python", "Fazer exercícios", "Enviar trabalho"]

tarefas.append("Lavar a louça")

tarefas.remove("Fazer exercícios")

print("Tarefas pendentes:", tarefas)
```
