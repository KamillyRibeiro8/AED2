Exercício 1: Crie uma lista em Python chamada frutas com os seguintes elementos: "maçã", "banana",
"laranja". Em seguida, adicione a fruta "morango" ao final da lista. Por fim, imprima a lista completa.

```python
frutas = ["maçã", "banana", "laranja"]
frutas.append("morango")
print("Frutas:", frutas)
```

Exercício 2: Crie uma pilha (usando uma lista Python) chamada livros. Adicione os livros "O Pequeno
Príncipe", "Dom Quixote" e "1984" nesta ordem. Em seguida, simule a remoção do livro do topo da pilha
e imprima o livro removido e a pilha restante.


```python
livros = ["O Pequeno Principe", "Dom Quixote", "1984"]
item_removido = livros.pop(0)
print("Livro removido: ", {item_removido} )
print("Livros restantes: ", livros)
```

Exercício 3: Crie uma fila (usando uma lista Python) chamada clientes. Adicione os clientes "Ana",
"Bruno" e "Carla" nesta ordem. Em seguida, simule a chegada de um novo cliente "Daniel" e a saída do
primeiro cliente da fila. Imprima o cliente que saiu e a fila restante.


```python
chamada_clientes = ["Ana", "Bruno", "Carla"]
chamada_clientes.append("Daniel")
cliente_removido = chamada_clientes.pop(0)
print("Cliente removido: ", cliente_removido)
print("Clientes presentes: ", chamada_clientes)
```
