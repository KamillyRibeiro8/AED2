5. (Prática) Implemente em Python uma função decimal_para_hexa(n) que converta um número decimal não-
negativo  n  para  sua  representação  hexadecimal  como  string,  usando  uma  pilha.  (Por  exemplo, 
decimal_para_hexa(26) deve retornar "1A".) 
 


```python
def decimal_para_hexa(n):
    if n == 0:
        return "0"
    
    hex_chars = "0123456789ABCDEF"
    pilha = []
    
    while n > 0:
        resto = n % 16
        pilha.append(hex_chars[resto])
        n //= 16
        
    resultado = ""
    while pilha:
        resultado += pilha.pop()
        
    return resultado

print(decimal_para_hexa(26))  
```

6. (Prática) Escreva uma função balanceia_colchetes(expressao) que, dada uma string contendo (, ), {, } e outros 
caracteres, determine se os parênteses e chaves estão balanceados. Use pilhas no seu algoritmo. Por exemplo, 
"({[]})" está balanceado, mas "({[}])" não está.

```python
def balanceia_colchetes(expressao):
    pilha = []
    mapeamento = {')': '(', '}': '}', ']': '['}
    
    for char in expressao:
        if char in "({[":
            pilha.append(char)
        elif char in ")}]":
            if not pilha or pilha[-1] != mapeamento[char]:
                return False
            pilha.pop()
            
    return len(pilha) == 0

print(balanceia_colchetes("({[]})"))  
print(balanceia_colchetes("({[}])"))  
```

# Projeto 1 – Histórico de Filtros em Dashboard de Dados 

Contexto: Em aplicações de ciência de dados para negócios, usuários interagem com dashboards aplicando 
filtros e transformações aos dados (por exemplo, em tabelas ou gráficos interativos). É útil permitir que o usuário desfaça filtros aplicados (por exemplo, revertendo ao estado anterior dos dados) sem recarregar tudo do início. 

Problema: Como implementar um mecanismo de histórico de operações que guarde estados anteriores dos 
dados ao aplicar filtros, para que o usuário possa desfazer a última ação? 
 


```python
class HistoricoDashboard:
    def __init__(self, dados_iniciais):
        self.dados_atuais = dados_iniciais
        self.pilha_historico = []

    def aplicar_filtro(self, novos_dados):
        # Guarda o estado atual na pilha antes de alterar
        self.pilha_historico.append(self.dados_atuais)
        self.dados_atuais = novos_dados

    def desfazer(self):
        if not self.pilha_historico:
            print("Nenhum filtro para desfazer.")
            return
        # Restaura o último estado guardado
        self.dados_atuais = self.pilha_historico.pop()

# Exemplo de uso
dash = HistoricoDashboard(["Dado 1", "Dado 2", "Dado 3"])
dash.aplicar_filtro(["Dado 1", "Dado 2"])  # Filtrou
dash.desfazer()                            # Volta ao estado inicial
```

# Projeto 2 – Calculadora de Expressões Aritméticas 

Contexto: Muitas calculadoras e programas de matemática interpretam expressões aritméticas infixas (como 
"3 + 4 * (2 - 1)"). Uma maneira clássica de avaliar tais expressões é converter para notação pós-fixa (RPN) ou usar pilhas para operar em tempo de execução. 

Problema: Escrever um programa Python que avalie uma expressão matemática em infixo contendo +, -, *, /, e 
parênteses. Queremos usar pilhas para lidar com a precedência de operadores e a aninhamento de parênteses. 

```python
def avalia_posfixa(expressao_posfixa):
    pilha = []
    tokens = expressao_posfixa.split()

    for token in tokens:
        if token.isdigit():
            pilha.append(float(token))
        else:
            b = pilha.pop()
            a = pilha.pop()
            if token == '+': pilha.append(a + b)
            elif token == '-': pilha.append(a - b)
            elif token == '*': pilha.append(a * b)
            elif token == '/': pilha.append(a / b)

    return pilha.pop()

print(avalia_posfixa("3 4 2 * +"))  
```
