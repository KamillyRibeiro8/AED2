5.  (PRÁTICA):  Classe  Pessoa:  Crie  uma  classe  Pessoa  com  nome  e  idade.  Adicione  método 
faz_aniversario() que incrementa a idade em 1. Use __str__() para retornar algo como "<nome>, <idade> 
anos". Crie duas pessoas e mostre a idade de cada uma antes e depois de chamar faz_aniversario(). 


```python
class Pessoa:

    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    def faz_aniversario(self):
        self.idade += 1

    def __str__(self):
        return f"{self.nome}, {self.idade} anos"


# Criando duas pessoas
p1 = Pessoa("Ana", 20)
p2 = Pessoa("Bruno", 25)

# Idade antes do aniversário
print("Antes do aniversário:")
print(p1)
print(p2)

# Chamando faz_aniversario()
p1.faz_aniversario()
p2.faz_aniversario()

# Idade depois do aniversário
print("\nDepois do aniversário:")
print(p1)
print(p2)
```

6.  (PRÁTICA):  Encapsulamento:  Dada  a  classe  abaixo  sem  encapsulamento,  implemente 
encapsulamento adequado: 
class Lampada:

 def __init__(self):

 self.estado = False

 def ligar(self):

 self.estado = True

 def desligar(self):

 self.estado = False

 Modifique para que o atributo estado seja privado e só possa ser alterado pelos métodos ligar() e
desligar(). Adicione métodos is_acesa() para verificar o estado.


```python
class Lampada:

    def __init__(self):
        self.__estado = False  # Atributo privado usando dois underscores (__)

    def ligar(self):
        self.__estado = True

    def desligar(self):
        self.__estado = False

    def is_acesa(self):
        return self.__estado


# Teste da classe
l = Lampada()
print(f"A lâmpada está acesa? {l.is_acesa()}")  # False

l.ligar()
print(f"A lâmpada está acesa? {l.is_acesa()}")  # True

l.desligar()
print(f"A lâmpada está acesa? {l.is_acesa()}")  # False
```

7. (PRÁTICA): Herança: Implemente duas classes com herança:

• Classe base Animal com um método fazer_som() que imprime algo genérico, por exemplo
"Algum som".

• Duas subclasses Cachorro e Gato que herdam de Animal. Na classe Cachorro, sobrescreva
fazer_som() para imprimir "Au Au". Em Gato, para "Miau".
Crie um objeto de cada e chame fazer_som() em cada um.


```python
class Animal:

    def fazer_som(self):
        print("Algum som")


class Cachorro(Animal):

    def fazer_som(self):
        print("Au Au")


class Gato(Animal):

    def fazer_som(self):
        print("Miau")


# Criando os objetos
cao = Cachorro()
gato = Gato()

# Chamando o método em cada um
cao.fazer_som()
gato.fazer_som()
```

8. (PRÁTICA): Polimorfismo: Dada as classes do Exercício 4, escreva uma função externa
animais_falam(lista_animais) que recebe uma lista de objetos (que podem ser Cachorro ou Gato) e
chama fazer_som() em cada um. Mostre que a função pode operar sobre diferentes tipos de animais
(polimorfismo em ação).

```python
def animais_falam(lista_animais):
    for animal in lista_animais:
        animal.fazer_som()


# Criando uma lista com objetos de classes diferentes
meus_animais = [Cachorro(), Gato(), Cachorro(), Gato()]

# A função executa o fazer_som() apropriado para cada tipo de animal
animais_falam(meus_animais)
```

# Projeto 1: Biblioteca de Livros
Contexto do mundo real: Imagine um sistema para gerenciar o empréstimo de livros em uma pequena
biblioteca escolar. Queremos cadastrar livros e usuários, registrar quando alguém empresta ou devolve um
livro.

Descrição do problema: Cada Livro tem título, autor, quantidade disponível. Cada Usuário tem nome,
matrícula e lista de livros emprestados. O sistema deve permitir que um usuário faça empréstimos (reduzindo
a quantidade disponível) e devoluções (aumentando quantidade). Devemos evitar empréstimos se o livro
estiver indisponível. Também precisamos manter os dados organizados e seguros (ex.: não permitir mexer
diretamente na quantidade disponível).

```python
class Livro:

    def __init__(self, titulo, autor, quantidade):
        self.titulo = titulo
        self.autor = autor
        self.__quantidade = quantidade  # Atributo privado para controle seguro

    def get_quantidade(self):
        return self.__quantidade

    def emprestar(self):
        if self.__quantidade > 0:
            self.__quantidade -= 1
            return True
        return False

    def devolver(self):
        self.__quantidade += 1


class Usuario:

    def __init__(self, nome, matricula):
        self.nome = nome
        self.matricula = matricula
        self.livros_emprestados = []

    def pegar_emprestado(self, livro):
        if livro.emprestar():
            self.livros_emprestados.append(livro.titulo)
            print(f"{self.nome} pegou '{livro.titulo}' emprestado.")
        else:
            print(
                f"Não foi possível emprestar '{livro.titulo}': Livro indisponível."
            )

    def devolver_livro(self, livro):
        if livro.titulo in self.livros_emprestados:
            livro.devolver()
            self.livros_emprestados.remove(livro.titulo)
            print(f"{self.nome} devolveu '{livro.titulo}'.")
        else:
            print(f"{self.nome} não possui o livro '{livro.titulo}'.")


# Teste do Projeto 1
livro1 = Livro("O Senhor dos Anéis", "J.R.R. Tolkien", 1)
u1 = Usuario("Lucas", "202301")

u1.pegar_emprestado(livro1)  # Sucesso
u1.pegar_emprestado(livro1)  # Falha (indisponível)
u1.devolver_livro(livro1)  # Sucesso
```


# Projeto 2: Gerenciamento de Conta Bancária com Herança

Contexto do mundo real: Um banco que oferece dois tipos de conta: conta corrente e conta poupança.
Queremos um sistema que permita depositar, sacar e aplicar juros (no caso da poupança).

Descrição do problema: Precisamos modelar contas bancárias em Python de forma orientada a objetos.
Teremos:

• Classe base Conta com atributos comuns (titular, saldo) e métodos depositar(valor) e sacar(valor).

• Classe ContaCorrente que herda de Conta e adiciona, por exemplo, taxa de saque (ou limite
especial).

• Classe ContaPoupanca que herda de Conta e tem um método extra render_juros() que aumenta o
saldo de acordo com uma taxa fixa anual.
Também podemos manter atributos de classe (como juros = 0.05 para 5% ao ano).

```python
class Conta:

    def __init__(self, titular, saldo_inicial=0.0):
        self.titular = titular
        self.saldo = saldo_inicial
 
    def depositar(self, valor):
        if valor > 0:
            self.saldo += valor
            print(f"Depósito de R$ {valor:.2f} realizado na conta de {self.titular}.")

    def sacar(self, valor):
        if 0 < valor <= self.saldo:
            self.saldo -= valor
            print(f"Saque de R$ {valor:.2f} realizado na conta de {self.titular}.")
            return True
        else:
            print(f"Saldo insuficiente para realizar o saque de R$ {valor:.2f}.")
            return False


class ContaCorrente(Conta):

    def __init__(self, titular, saldo_inicial=0.0, taxa_saque=2.50):
        super().__init__(titular, saldo_inicial)
        self.taxa_saque = taxa_saque

    def sacar(self, valor):
        valor_total = valor + self.taxa_saque
        print(f"Taxa de saque aplicada: R$ {self.taxa_saque:.2f}")
        return super().sacar(valor_total)


class ContaPoupanca(Conta):
    taxa_juros = 0.05  # Atributo de classe (5% ao ano)

    def render_juros(self):
        juros = self.saldo * ContaPoupanca.taxa_juros
        self.saldo += juros
        print(f"Juros de R$ {juros:.2f} aplicados à conta de {self.titular}.")


# Teste do Projeto 2
cc = ContaCorrente("Maria", 100.00)
cc.sacar(20.00)  # Desconta 20.00 + 2.50 de taxa
print(f"Saldo atual CC: R$ {cc.saldo:.2f}")

cp = ContaPoupanca("João", 500.00)
cp.render_juros()  # Rende 5% de 500 = +25
print(f"Saldo atual CP: R$ {cp.saldo:.2f}")
```
